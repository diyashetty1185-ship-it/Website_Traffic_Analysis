```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/website_traffic_2025.csv")
```

```python
df.head()
```

```python
#Univariate Analysis
df['Total_Visitors'].describe()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Total_Visitors'], bins=30, kde=True)
plt.title("Distribution of Total Visitors")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['Revenue'])
plt.title("Revenue Spread")
plt.show()
```

```python
#Bivariate Analysis
plt.figure(figsize=(8,5))
sns.scatterplot(x='Total_Visitors', y='Revenue', data=df, alpha=0.6)
plt.title("Visitors vs Revenue")
plt.show()
```

```python
sns.lmplot(x='Bounce_Rate', y='Conversion_Rate', data=df, aspect=1.5)
plt.title("Bounce Rate vs Conversion Rate")
plt.show()
```

```python
df.groupby('Device_Type')['Avg_Time_Spent'].mean()
```

```python
#Multivariate Analysis
# Correlation heatmap
plt.figure(figsize=(10,8))
sns.heatmap(df.select_dtypes(include='number').corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#Pairplot
sns.pairplot(df)
```

```python
sns.scatterplot(x='Total_Visitors',
                y='Revenue',
                hue='Conversion_Rate',
                data=df)
```
